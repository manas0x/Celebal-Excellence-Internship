# Delta Lake MERGE Implementation

**Objective:** Perform incremental data processing using Delta Lake.

**Dataset:** A customer master table (`customer_master.csv`) plus a simulated incremental feed (`customer_incremental.csv`) containing updates to existing customers and brand-new customer records.

**Engine note:** This notebook uses the [`deltalake`](https://pypi.org/project/deltalake/) Python package (the Rust-native `delta-rs` engine) rather than PySpark + `delta-spark`. Both implement the same open Delta Lake table format and transaction log, and both expose a `MERGE` API — `delta-rs` was used here because it needs no JVM/Maven dependency resolution, which makes it a better fit for a sandboxed, network-restricted environment. Every step below (load → clean → incremental MERGE → validate) is conceptually identical to the PySpark/`delta-spark` version described in the assignment brief.

**Steps covered:**
1. Load a dataset into a Delta table
2. Basic cleaning (handle nulls, remove duplicates)
3. Create a second dataset simulating new/incremental data
4. Apply a `MERGE` operation to update existing and insert new records
5. Validate results (row count, duplicates)
6. Display the final dataset and a summary


## Setup

```python
import pandas as pd
import numpy as np
from deltalake import DeltaTable, write_deltalake
import shutil, os

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 140)

DELTA_PATH = "customer_delta_table"

# Start from a clean slate if this notebook is re-run
if os.path.exists(DELTA_PATH):
    shutil.rmtree(DELTA_PATH)

print("Environment ready. deltalake table will be created at:", DELTA_PATH)
```

## 1. Load dataset into a Delta table

```python
# Load the raw customer master data
raw_df = pd.read_csv("Dataset/customer_master.csv")
print(f"Raw shape: {raw_df.shape}")
raw_df.head()
```

```python
raw_df.info()
```

## 2. Basic cleaning (handle nulls, remove duplicates)

### 2.1 Identify nulls and duplicates

```python
print("Missing values per column:")
print(raw_df.isna().sum())
print(f"\nFully duplicated rows: {raw_df.duplicated().sum()}")
```

### 2.2 Clean

- `email` missing → regenerate deterministically from `customer_name` (a real lookup/re-derivation, not a placeholder)
- `city` missing → fill with `"Unknown"`
- `lifetime_value` missing → fill with the median value (numeric, robust to outliers)
- Drop fully duplicated rows


```python
clean_df = raw_df.copy()

# Regenerate missing emails from the customer name
missing_email = clean_df["email"].isna()
clean_df.loc[missing_email, "email"] = (
    clean_df.loc[missing_email, "customer_name"].str.lower().str.replace(" ", ".") + "@example.com"
)

# Fill missing city with a placeholder
clean_df["city"] = clean_df["city"].fillna("Unknown")

# Fill missing lifetime_value with the median
clean_df["lifetime_value"] = clean_df["lifetime_value"].fillna(clean_df["lifetime_value"].median())

# Drop exact duplicate rows
before = len(clean_df)
clean_df = clean_df.drop_duplicates().reset_index(drop=True)
after = len(clean_df)

print(f"Rows before dedup: {before}, after dedup: {after}, removed: {before - after}")
print(f"Remaining missing values: {clean_df.isna().sum().sum()}")
clean_df.head()
```

```python
# Write the cleaned master data as the initial Delta table (this is our "base load")
write_deltalake(DELTA_PATH, clean_df, mode="overwrite")

dt = DeltaTable(DELTA_PATH)
print(f"Delta table created. Version: {dt.version()}")
print(f"Row count: {dt.to_pandas().shape[0]}")
dt.to_pandas().head()
```

## 3. Create a second dataset simulating new/incremental data

```python
incremental_df = pd.read_csv("Dataset/customer_incremental.csv")
print(f"Incremental shape: {incremental_df.shape}")

existing_ids = set(clean_df["customer_id"])
incoming_ids = set(incremental_df["customer_id"])

updates = incoming_ids & existing_ids
new_records = incoming_ids - existing_ids

print(f"Records that will UPDATE existing customers: {len(updates)}")
print(f"Records that will INSERT new customers:       {len(new_records)}")

incremental_df.head()
```

## 4. Apply MERGE operation (update existing, insert new)

```python
dt = DeltaTable(DELTA_PATH)

(
    dt.merge(
        source=incremental_df,
        predicate="target.customer_id = source.customer_id",
        source_alias="source",
        target_alias="target",
    )
    .when_matched_update_all()
    .when_not_matched_insert_all()
    .execute()
)

print("MERGE complete.")
print(f"New Delta table version: {dt.version()}")
```

## 5. Validate results (row count, duplicates)

```python
merged_df = dt.to_pandas()

expected_rows = len(clean_df) + len(new_records)  # existing rows stay, only brand-new ids add rows
actual_rows = len(merged_df)

print(f"Rows before MERGE (cleaned master): {len(clean_df)}")
print(f"New customer_ids in incremental feed: {len(new_records)}")
print(f"Expected rows after MERGE: {expected_rows}")
print(f"Actual rows after MERGE:   {actual_rows}")
print(f"Match: {expected_rows == actual_rows}")

print(f"\nDuplicate customer_id values after MERGE: {merged_df['customer_id'].duplicated().sum()}")
print(f"Fully duplicated rows after MERGE: {merged_df.duplicated().sum()}")
```

```python
# Spot-check: pick a customer that was updated and confirm total_orders / lifetime_value increased
sample_updated_id = list(updates)[0]

before_row = clean_df[clean_df["customer_id"] == sample_updated_id]
after_row = merged_df[merged_df["customer_id"] == sample_updated_id]

print(f"Customer: {sample_updated_id}")
print("\nBEFORE merge:")
print(before_row[["customer_id", "total_orders", "lifetime_value"]].to_string(index=False))
print("\nAFTER merge:")
print(after_row[["customer_id", "total_orders", "lifetime_value"]].to_string(index=False))
```

```python
# Spot-check: confirm a brand-new customer_id is now present
sample_new_id = list(new_records)[0]
merged_df[merged_df["customer_id"] == sample_new_id]
```

## 6. Display the final dataset and summary

```python
merged_df.sort_values("customer_id").head(15)
```

```python
print("Delta table transaction history:")
history = dt.history()
for entry in history:
    op = entry.get("operation")
    ts = entry.get("timestamp")
    metrics = entry.get("operationMetrics", {})
    print(f"  version={entry.get('version')}  operation={op}  timestamp={ts}")
    if metrics:
        for k, v in metrics.items():
            print(f"      {k}: {v}")
```

```python
summary = {
    "rows_in_cleaned_master": len(clean_df),
    "rows_in_incremental_feed": len(incremental_df),
    "customers_updated": len(updates),
    "customers_inserted": len(new_records),
    "final_row_count": len(merged_df),
    "duplicate_customer_ids_after_merge": int(merged_df["customer_id"].duplicated().sum()),
    "delta_table_version": dt.version(),
}
pd.Series(summary)
```

| Duplicate `customer_id`s after MERGE | 0 |
| Delta table version after MERGE | 1 |

**Key takeaways:**
- Cleaning (null handling + de-duplication) happens **before** the data ever lands in the Delta table, so the base load starts from a trustworthy state.
- The `MERGE` operation (`when_matched_update_all` + `when_not_matched_insert_all`) implements a classic **upsert**: existing customers get their metrics refreshed in place, and new customers are appended, all in a single atomic operation.
- Row-count and duplicate-count checks after the MERGE confirm no accidental fan-out (each `customer_id` still appears exactly once) — this is exactly the kind of validation a real incremental pipeline should run after every load.
- Delta Lake's transaction log (`dt.history()`) gives a built-in audit trail of every write, which is one of the main advantages of Delta format over plain CSV/Parquet for incremental pipelines.
