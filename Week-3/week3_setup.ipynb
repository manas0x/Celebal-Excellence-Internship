```python
%pip install kagglehub

import kagglehub
import shutil
from pathlib import Path

# Download latest version from KaggleHub
path = Path(kagglehub.dataset_download("vivek468/superstore-dataset-final"))

# Current working directory (Week-3 folder)
current_dir = Path.cwd()

# Create DataSet folder if it doesn't exist
dataset_dir = current_dir / "DataSet"

# Source CSV in KaggleHub cache
file = path / "Sample - Superstore.csv"

# Copy CSV to Week-3/DataSet/
shutil.copy(file, dataset_dir / file.name)

print(f"Dataset copied successfully to: {dataset_dir / file.name}")
```

```python
!pip3 install pandas
import pandas as pd

df = pd.read_csv("Sample - Superstore.csv" , encoding="latin1")
#CSV Encoding to latin1


df.to_csv("Sample - Superstore.csv", index=False, encoding="utf-8")
#CSV Encoding to UTF-8

print("Done!")

df.head()
df.shape
df.info()
```

```python
!pip install sqlalchemy

from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:manas@localhost/superstore"
)

df.to_sql(
    "superstore_raw",
    engine,
    if_exists="replace",
    index=False
)

print("Data inserted into MySQL database successfully!")
```
